```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import seaborn as sns


sns.set()
```

```python
data = pd.read_csv('Multiple+linear+regression.csv')
data.head()
```

```python
data.describe()
```

## Create the multiple linear regression

### Declare the dependent and independent variables

```python
y = data['GPA']
x = data[['SAT', 'Rand 1,2,3']]
```

## Regression itself

```python
reg = LinearRegression()
```

```python
reg.fit(x, y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

## Calculating R-squared

```python
reg.score(x,y) #the R-squared itself
```

### Formula for Adjusted R^2
$R^2_{adj.} = 1 - (1-R^2)*\frac{n-1}{n-p-1}$

```python
x.shape #84 is the number of observations, 2 is the number of predictors
```

```python
r2 = reg.score(x,y)
n = x.shape[0]
p = x.shape[1]

adjusted_r2 = 1 - (1-r2) * (n-1) / (n-p-1)
adjusted_r2
```

## Feature selection(F-regression)

It helps us to drop all unneeded features by:

```python
from sklearn.feature_selection import f_regression
```

```python
f_regression(x, y)
```

```python
p_values = f_regression(x, y)[1]
p_values
```

```python
p_values.round(3)
```

### Creating a summary table

```python
reg_summary = pd.DataFrame(data = x.columns.values, columns = ['Features'])
reg_summary
```

```python
reg_summary['Coefficients'] = reg.coef_
reg_summary['p-values'] = p_values.round(3)
```

```python
reg_summary
```
